# 通过FFMPEG启动RTSP

*FFMPEG* - 一组开源的免费库，允许以各种格式录制、转换和传输数字音频和视频。

*RTSP (实时流媒体协议)* - 用于控制音频和视频等数据流传输的网络协议。

### 通过ffmpeg启动RTSP流：
- `-i $VIDEO_PATH` - 视频文件路径(mp4/avi/mov)
- `$RTSP` - 流地址
- `-stream_loop -1` - 循环播放视频n次(-1表示无限循环)
- `-re` - 使用视频的原生FPS(帧率)
- `-f rtsp` - 通过RTSP协议进行流传输
- `-rtsp_transport tcp` - 选择TCP作为传输协议(替代方案是UDP)

```python
VIDEO_PATH="../test_video.mp4"    
RTSP="rtsp://127.0.0.1:8554/input"

! ffmpeg -re -stream_loop -1 -i $VIDEO_PATH  -f rtsp -rtsp_transport tcp $RTSP
```
